# Linux Commands Cheat Sheet for EC2 Deployments

This notebook covers the essential Linux/Unix commands you need for deploying a Streamlit app on EC2.
Each section includes the command and an explanation of what it does.

**Note:** Commands prefixed with `!` run in the system shell. These work in Jupyter on Linux/Mac.
On EC2, run these in your terminal (without the `!`).

## 1. Navigation — Where Am I?

Before doing anything, know where you are.

```python
# Print Working Directory — shows your current location
!pwd
```

```python
# Show current username
!whoami
```

```python
# Show OS version — important for choosing the right package manager
!cat /etc/os-release
```

## 2. Directory Navigation

```python
# These are run in your terminal on EC2:
# cd /opt/apps            → go to /opt/apps
# cd ..                   → go up one level
# cd ~                    → go to home directory
# cd -                    → go to previous directory
# mkdir -p /opt/apps/myapp → create directory (and parents)

# Demonstration (safe to run in notebook):
import os
print("Current directory:", os.getcwd())
```

## 3. Listing Files

```python
# ls          → list files
# ls -l       → long format (permissions, owner, size)
# ls -la      → long format including hidden files (like .env)
# ls -lah     → human-readable sizes (KB, MB)
# ls -ltr     → sorted by time, oldest first

!ls -lah .
```

## 4. Viewing File Contents

```python
# cat file.txt        → print entire file
# head file.txt       → first 10 lines
# tail file.txt       → last 10 lines
# tail -f app.log     → follow a log file in real time
# less file.txt       → page through file (q to quit)

# Example — view a requirements.txt
# !cat requirements.txt

# Simulate: show first 5 lines of /etc/hosts
!head -5 /etc/hosts
```

## 5. File Operations — Create, Copy, Move, Delete

> ⚠️ `rm -rf` is permanent. There is no recycle bin on Linux.

```python
# touch file.txt          → create empty file
# cp source dest          → copy file
# cp -r src_dir dest_dir  → copy directory
# mv old new              → rename or move
# rm file.txt             → delete file
# rm -r folder/           → delete folder
# rm -rf folder/          → force delete folder (no prompts)

# Safe demonstration:
!touch /tmp/test_file.txt
!ls -la /tmp/test_file.txt
!rm /tmp/test_file.txt
print("File created and removed successfully")
```

## 6. Searching

```python
# Find files by name:
# find . -name "app.py"                   → find app.py in current dir
# find / -name "python3.11" 2>/dev/null   → find anywhere, ignore errors

# Search text inside files:
# grep "streamlit" app.py                 → find word in one file
# grep -r "streamlit" .                   → search all files recursively
# grep -Rin "streamlit" .                 → recursive, case-insensitive, show line numbers

!find /etc -name "nginx.conf" 2>/dev/null
```

## 7. Processes and Ports

```python
# See running processes:
# ps -ef                      → all processes with details
# ps -ef | grep streamlit     → filter for streamlit
# top                         → live interactive process viewer

# Kill a process:
# kill 12345       → graceful stop by PID
# kill -9 12345    → force kill by PID
# pkill -f streamlit → kill by name

# Check listening ports:
# ss -tulpn                   → all listening ports
# ss -tulpn | grep 8501       → check streamlit port
# ss -tulpn | grep :80        → check nginx port

!ps -ef | grep python | grep -v grep | head -5
```

## 8. Permissions

```python
# Permissions format: -rwxrwxrwx
# Position 1: file type (- = file, d = directory)
# Positions 2-4: owner permissions (r=read, w=write, x=execute)
# Positions 5-7: group permissions
# Positions 8-10: others permissions

# chmod 600 .env     → owner: rw, group: none, others: none (for secret files)
# chmod 755 dir/     → owner: rwx, group: rx, others: rx (for directories)
# chmod +x script.sh → make executable

# chown user:group file  → change owner
# chown -R ec2-user:ec2-user /opt/apps/myapp  → recursive ownership

# Demonstration:
!touch /tmp/test_perms.txt
!chmod 600 /tmp/test_perms.txt
!ls -la /tmp/test_perms.txt
!rm /tmp/test_perms.txt
```

## 9. Disk and Memory

```python
# Disk usage:
# df -h              → disk space for all filesystems (human-readable)
# du -sh .           → size of current directory
# du -sh /opt/apps/* → size of each app

# Memory:
# free -h            → total, used, available RAM

!df -h /
print()
!free -h
```

## 10. systemd Service Commands

```python
# These run in your EC2 terminal (not in notebook — requires sudo):

commands = {
    "systemctl start streamlit-app":    "Start the Streamlit service",
    "systemctl stop streamlit-app":     "Stop the service",
    "systemctl restart streamlit-app":  "Restart (stop + start)",
    "systemctl enable streamlit-app":   "Enable auto-start on boot",
    "systemctl status streamlit-app":   "Show current status",
    "journalctl -u streamlit-app -f":   "Follow live service logs",
    "journalctl -u streamlit-app -n 100": "Last 100 log lines",
    "systemctl daemon-reload":          "Reload after editing service file",
}

print(f"{'Command':<45} {'What it does'}")
print("-" * 80)
for cmd, desc in commands.items():
    print(f"{cmd:<45} {desc}")
```

## 11. Nginx Commands

```python
nginx_commands = {
    "nginx -t":                         "Test config syntax — always run before reload",
    "systemctl reload nginx":           "Apply config changes (no downtime)",
    "systemctl restart nginx":          "Full restart",
    "systemctl status nginx":           "Check if Nginx is running",
    "tail -f /var/log/nginx/access.log": "Watch incoming requests",
    "tail -f /var/log/nginx/error.log": "Watch errors",
}

print(f"{'Command':<45} {'What it does'}")
print("-" * 80)
for cmd, desc in nginx_commands.items():
    print(f"{cmd:<45} {desc}")
```
